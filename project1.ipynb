```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    "Date": ["2026-01-05","2026-01-10","2026-02-03","2026-02-15",
             "2026-03-02","2026-03-18","2026-04-05","2026-04-20",
             "2026-05-10","2026-05-25"],
    "Store": ["Store A","Store B","Store A","Store C","Store B",
              "Store A","Store C","Store B","Store A","Store C"],
    "Product": ["Laptop","Mobile","Headphones","Laptop","Mobile",
                "Tablet","Headphones","Laptop","Mobile","Tablet"],
    "Category": ["Electronics","Electronics","Accessories","Electronics",
                 "Electronics","Electronics","Accessories","Electronics",
                 "Electronics","Electronics"],
    "Quantity": [2,5,8,3,6,4,10,2,7,5],
    "Sales": [100000,75000,24000,150000,90000,80000,30000,70000,105000,100000],
    "Profit": [15000,12000,5000,22000,15000,10000,7000,11000,18000,16000]
}

df = pd.DataFrame(data)
df
```

```python
print("First 5 rows:")
print(df.head())

print("\nDataset Information:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("Total Sales:", df["Sales"].sum())
print("Total Profit:", df["Profit"].sum())
print("Total Quantity Sold:", df["Quantity"].sum())
```

```python
store_sales = df.groupby("Store")["Sales"].sum().sort_values(ascending=False)

print("Sales by Store:")
print(store_sales)
```

```python
product_sales = df.groupby("Product")["Sales"].sum().sort_values(ascending=False)

print("Sales by Product:")
print(product_sales)
```

```python
category_sales = df.groupby("Category")["Sales"].sum()

print("Sales by Category:")
print(category_sales)
```

```python
store_sales.plot(kind="bar", figsize=(7,5))

plt.title("Store Performance")
plt.xlabel("Store")
plt.ylabel("Total Sales")
plt.show()
```

```python
product_sales.plot(kind="bar", figsize=(8,5))

plt.title("Product Sales")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
df["Date"] = pd.to_datetime(df["Date"])

daily_sales = df.groupby("Date")["Sales"].sum()

plt.figure(figsize=(10,5))
plt.plot(daily_sales.index, daily_sales.values, marker="o")

plt.title("Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.heatmap(
    df[["Quantity", "Sales", "Profit"]].corr(),
    annot=True,
    cmap="Blues"
)

plt.title("Sales Data Correlation")
plt.show()
```

```python
best_store = store_sales.idxmax()
best_product = product_sales.idxmax()

print("===== FINAL RESULT =====")
print("Total Sales:", df["Sales"].sum())
print("Total Profit:", df["Profit"].sum())
print("Best Performing Store:", best_store)
print("Best Selling Product:", best_product)
```
